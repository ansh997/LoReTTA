```python
import pandas as pd
import os

dataset_name = 'bitcoin'
path = rf'/raid/t2/TGN_adv/DyGlib/DG_data/{dataset_name}'

input_file = 'soc-sign-bitcoinotc.csv'
# output_file = 'CollegeMsg.csv'

# Assuming the TXT file is tab-delimited
df = pd.read_csv(os.path.join(path, input_file),
                #  delimiter=' ', header=None, names=['src', 'dst', 'time']
                )  # Adjust delimiter if needed

df.columns=['u', 'i', 'label', 'ts']
columns = df.columns.tolist()
columns[-1], columns[-2] = columns[-2], columns[-1]  # Swap the last two columns
df = df[columns]
df['idx'] = df.index
df['label'] = df['label'].astype(float)
df['idx'] = df['idx'].astype(int)
df.sort_values(by=['ts'], inplace=True)

df['ts'] = df['ts'].astype(int)
# df['ts'] = df['ts'] - df['ts'].iloc[0]

df.head()
```
