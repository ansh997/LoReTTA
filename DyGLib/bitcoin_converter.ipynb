```python
import pandas as pd

import os

dataset_name = 'bitcoin'
path = r'/raid/t2/TGN_adv/DyGlib/DG_data/{}/'.format(dataset_name)

save_filename = f'ml_{dataset_name}.csv'
load_filename = f'soc-sign-bitcoinotc.csv'

df = pd.read_csv(os.path.join(path, load_filename))
df.columns=['u', 'i', 'label', 'ts']
columns = df.columns.tolist()
columns[-1], columns[-2] = columns[-2], columns[-1]  # Swap the last two columns
df = df[columns]
df['idx'] = df.index
df['label'] = df['label'].astype(float)
df['idx'] = df['idx'].astype(int)
df.sort_values(by=['ts'], inplace=True)
df['ts'] = df['ts'].abs().astype(int)
df['ts'] = df['ts'] - df['ts'].iloc[0]
# df['ts'] = df['ts'].astype(float)
df
```
